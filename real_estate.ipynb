```python
outlier_cols = [
    "BHK",
    "Size_in_SqFt",
    "Price_in_Lakhs",
    "Price_per_SqFt",
    "Year_Built",
    "Floor_No",
    "Total_Floors",
    "Age_of_Property",
    "Nearby_Schools",
    "Nearby_Hospitals"
]

for col in outlier_cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    count = ((df[col] < lower) | (df[col] > upper)).sum()

    print(f"{col}: {count} outliers")
```

```python
df["Amenity_Density_Score"] = (
    df["Amenities"]
    .fillna("")
    .apply(
        lambda x: len([item for item in str(x).split(",") if item.strip()])
    )
)
```

```python
categorical_check_cols = [
    "State",
    "City",
    "Locality",
    "Property_Type",
    "Furnished_Status",
    "Public_Transport_Accessibility",
    "Parking_Space",
    "Security",
    "Amenities",
    "Facing",
    "Owner_Type",
    "Availability_Status"
]

for col in categorical_check_cols:
    print(f"\n{col}:")
    print(df[col].value_counts().head(20))
```

```python
df["Parking_Space"] = df["Parking_Space"].map({
    "No": 0,
    "Yes": 1
})

df["Security"] = df["Security"].map({
    "No": 0,
    "Yes": 1
})

mapping = {
    "Low": 0,
    "Medium": 1,
    "High": 2
}

df["Public_Transport_Accessibility"] = (
    df["Public_Transport_Accessibility"].map(mapping)
)
```

```python
price_median = df["Price_per_SqFt"].median()
age_median = df["Age_of_Property"].median()
amenity_median = df["Amenity_Density_Score"].median()
school_median = df["Nearby_Schools"].median()
hospital_median = df["Nearby_Hospitals"].median()

df["Investment_Score"] = 0


# 1. Price per SqFt
df["Investment_Score"] += (
    df["Price_per_SqFt"] <= price_median
).astype(int)


# 2. Property Age
df["Investment_Score"] += (
    df["Age_of_Property"] <= age_median
).astype(int)


# 3. Amenities
df["Investment_Score"] += (
    df["Amenity_Density_Score"] >= amenity_median
).astype(int)


# 4. Public Transport
df["Investment_Score"] += (
    df["Public_Transport_Accessibility"] >= 1
).astype(int)


# 5. Security
df["Investment_Score"] += (
    df["Security"] >= 1
).astype(int)


# 6. Nearby Schools
df["Investment_Score"] += (
    df["Nearby_Schools"] >= school_median
).astype(int)


# 7. Nearby Hospitals
df["Investment_Score"] += (
    df["Nearby_Hospitals"] >= hospital_median
).astype(int)


# 8. Parking
df["Investment_Score"] += (
    df["Parking_Space"] == 1
).astype(int)

# 9. Furnished Status
df["Investment_Score"] += (
    df["Furnished_Status"].isin(["Furnished", "Semi-furnished"])
).astype(int)


# Final Good Investment classification
df["Good_Investment"] = (
    df["Investment_Score"] >= 3
).astype(int)

print("Investment Score Distribution:")
print(df["Investment_Score"].value_counts().sort_index())

print("\nGood Investment Distribution:")
print(df["Good_Investment"].value_counts())
```

```python
df["Good_Investment"] = df["Good_Investment"].map({1: "Yes", 0: "No"})
```

```python
amenities_clean = (
    df["Amenities"]
    .fillna("")
    .astype(str)
    .str.split(",")
    .apply(
        lambda x: [
            item.strip().lower().replace(" ", "_")
            for item in x
            if item.strip()
        ]
    )
)
```

## App metadata
Saves dropdown options, locality frequencies and defaults so that `app.py` can rebuild the exact training-time feature vector.

```python
import json
import os

os.makedirs("saved_models", exist_ok=True)

categorical_for_app = [
    "State", "City", "Property_Type", "Furnished_Status",
    "Facing", "Owner_Type", "Availability_Status"
]

app_metadata = {
    "categories": {
        col: sorted(df[col].dropna().unique().tolist())
        for col in categorical_for_app
    },
    "city_to_state": df.groupby("City")["State"].first().to_dict(),
    "locality_frequency": df["Locality"].value_counts().to_dict(),
    "amenities": unique_amenities,
    "reference_year": int((df["Year_Built"] + df["Age_of_Property"]).mode()[0]),
    "defaults": {
        col: float(df[col].median())
        for col in [
            "BHK", "Size_in_SqFt", "Price_in_Lakhs", "Year_Built",
            "Floor_No", "Total_Floors", "Nearby_Schools", "Nearby_Hospitals"
        ]
    },
    "ranges": {
        col: [float(df[col].min()), float(df[col].max())]
        for col in [
            "BHK", "Size_in_SqFt", "Price_in_Lakhs", "Year_Built",
            "Floor_No", "Total_Floors", "Nearby_Schools", "Nearby_Hospitals"
        ]
    },
}

with open("saved_models/app_metadata.json", "w") as f:
    json.dump(app_metadata, f, indent=2)

print("App metadata saved:", list(app_metadata.keys()))
```
